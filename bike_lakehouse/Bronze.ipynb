# Define ingestion configuration

```python
INGESTION_CONFIG=[
    {
       "path":"/Volumes/workspace/bronze/source_systems/cust_info.csv",
       "table":"cust_info"
    },
    {
        "path":"/Volumes/workspace/bronze/source_systems/prd_info.csv",
       "table":"prd_info"
    },
    {
        "path":"/Volumes/workspace/bronze/source_systems/sales_details.csv",
       "table":"sales_details"
    },
    {
        "path":"/Volumes/workspace/bronze/source_systems/CUST_AZ12.csv",
       "table":"CUST_AZ12"
    },
    {
        "path":"/Volumes/workspace/bronze/source_systems/LOC_A101.csv",
       "table":"LOC_A101"  
       
    },
    {
        "path":"/Volumes/workspace/bronze/source_systems/PX_CAT_G1V2.csv",
       "table":"PX_CAT_G1V2"
    }
]
```

# Ingest file into Bronze tables

```python
%python
for item in INGESTION_CONFIG:
    df = (
        spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load(item["path"])
    )

    df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(item["table"])
```
